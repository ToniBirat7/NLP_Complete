# **Basics**

## **Tokenization**

Tokenization is the process of breaking down text into smaller units called tokens. These tokens can be words, subwords, or characters, depending on the tokenization method used.

For example, the sentence "I love natural language processing!" can be tokenized into:

- Word-level tokens: ["I", "love", "natural", "language", "processing!"]

There are many tokenization methods, including:

**Word-level tokenization**: This method breaks text into individual words. It is simple but can lead to issues with out-of-vocabulary words.

**Subword tokenization**: This method breaks words into smaller units, such as Byte Pair Encoding (BPE) or WordPiece. It helps to handle out-of-vocabulary words and can capture morphological information.

**Character-level tokenization**: This method breaks text into individual characters. It can capture fine-grained information but may require longer sequences and more computational resources.

Currently, `Subword Tokenization` or `Byte-Pair Encoding (BPE)` is the most commonly used method in language modeling, as it strikes a balance between handling out-of-vocabulary words and capturing morphological information.

Apart from theses `Tokenization` methods, `Tokenizer-Free` approaches are also being explored, which do not rely on  explicit tokenization and instead `Bytes` are used as the basic units for processing text. This can help to avoid issues with out-of-vocabulary words and can capture more fine-grained information, but it may require more computational resources and longer sequences.
