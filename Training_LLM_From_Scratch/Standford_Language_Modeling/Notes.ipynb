## **Key Takeaways**



### **Some Design Decisions are simply not (yet) justifiable and just come from experimentation.**

- Example: The choice of activation function (ReLU, GELU, etc.) is often based on empirical results rather than theoretical justification.

- Example: The architecture of transformer models, such as the number of layers and attention heads, is often determined through trial and error.

- Example: Noam Shazeer paper that introduced `SwiGLU` activation function showed that it outperformed ReLU and GELU in certain scenarios, but the exact reasons for its effectiveness are still not fully understood.
