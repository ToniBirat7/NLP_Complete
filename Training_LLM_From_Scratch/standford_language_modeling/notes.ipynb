# **Basics**

## **Tokenization**

Tokenization is the process of breaking down text into smaller units called tokens. These tokens can be words, subwords, or characters, depending on the tokenization method used.

For example, the sentence "I love natural language processing!" can be tokenized into:

- Word-level tokens: ["I", "love", "natural", "language", "processing!"]

There are many tokenization methods, including:

**Word-level tokenization**: This method breaks text into individual words. It is simple but can lead to issues with out-of-vocabulary words.

**Subword tokenization**: This method breaks words into smaller units, such as Byte Pair Encoding (BPE) or WordPiece. It helps to handle out-of-vocabulary words and can capture morphological information.

**Character-level tokenization**: This method breaks text into individual characters. It can capture fine-grained information but may require longer sequences and more computational resources.

Currently, `Subword Tokenization` or `Byte-Pair Encoding (BPE)` is the most commonly used method in language modeling, as it strikes a balance between handling out-of-vocabulary words and capturing morphological information.

Apart from theses `Tokenization` methods, `Tokenizer-Free` approaches are also being explored, which do not rely on  explicit tokenization and instead `Bytes` are used as the basic units for processing text. This can help to avoid issues with out-of-vocabulary words and can capture more fine-grained information, but it may require more computational resources and longer sequences. ## **Model Architecture**

Once we've `Tokenized` the text into a sequence of tokens or integers, we now define the `Model Architecture` that will process this sequence.

The most common architecture for language modeling is the `Transformer` architecture [Vaswani et al. in 2017](https://arxiv.org/abs/2105.13626), which consists of multiple layers of self-attention and feedforward neural networks. The `Transformer` architecture has been shown to be effective for a wide range of natural language processing tasks, including language modeling.

Since 2017, there have been many variations and improvements to the original `Transformer` architecture, such as `BERT`, `GPT`, `T5`, and many others. These models have achieved state-of-the-art performance on various NLP tasks and have been widely adopted in both research and industry.

But at its core, the `Transformer` architecture remains the foundation for most modern language models, and it continues to be an active area of research and development in the field of natural language processing.

Also, there are improvements which made a substantial difference in the performance of language models such as:

- **Non Linear, Activation Functions**: The choice of activation function can have a significant impact on the performance of a language model. For example, the `SwiGLU` activation function introduced by Noam Shazeer has been shown to outperform traditional activation functions like `ReLU` and `GELU` in certain scenarios.

- **Positional Encodings**: The way positional information is encoded in the model can also affect its performance. For example, `Rotary Positional Encodings (RoPE)` have been shown to be more effective than traditional `Sinusoidal Positional Encodings` in certain cases.

- **Normalization Techniques**: The choice of normalization technique can also impact the performance of a language model. For example, `RMSNorm` has been shown to be more effective than `LayerNorm` in certain scenarios.

- **Placement of Normalization**: The placement of normalization layers within the model architecture can also affect performance. For example, `Pre-LN` (pre-layer normalization) has been shown to be more effective than `Post-LN` (post-layer normalization) in certain cases.

- **Multi Layer Perceptron or Feedforward Networks**: The design of the feedforward networks within the transformer architecture can also impact performance. The original transformer used `Dense` layers, but in current scenarios `Mixture of Experts (MoE)` layers have been shown to be more effective in certain cases.

- **Attention Mechanisms**: There are multiple variations of attentions such as `Full Attention`, `Sliding Window Attention`, `Linear Attention`, `Sparse Attention`, are used to prevent the quadratic growth of attention with respect to sequence length, and they can have a significant impact on the performance of a language model.

- **Lower-Dimensional Attention**: The use of lower-dimensional attention mechanisms, such as `Group Query Attention (GQA)`, `Multi-Head Latent Attention (MLA)`, and `Multi-Head Low-Rank Attention (MLRA)`, can help to reduce the computational complexity of attention while still capturing important dependencies in the data.

- **State Space Models**: State space models, such as `S4`, have been shown to be effective for modeling long-range dependencies in sequential data, and they can be used as an alternative to traditional transformer architectures in certain scenarios.

## **Training**

Once we have defined our model architecture, we need to train it on a large corpus of text data. This involves optimizing the model's parameters to minimize the loss function, which measures the difference between the model's predictions and the actual target values.

There are various things that can impact the training of a language model, such as:

- Optimizer: `AdamW`, `Muon`, `SOAP`, etc.

- Learning Rate Schedulers: `Cosine Annealing`, `WSD`

- Batch Size: The choice of batch size can impact the convergence and generalization of the model.

- Regularization Techniques: Techniques such as `Dropout`, `Weight Decay`, and `Gradient Clipping` can help to prevent overfitting and improve the generalization of the model.

- Hyperparameter Tuning: The choice of hyperparameters, such as the `learning rate`, `batch size`, `Number of heads` and model architecture, can have a significant impact on the performance of the language model. Hyperparameter tuning is often done through a combination of grid search, random search, and more advanced techniques like Bayesian optimization.
