## **Tokenization**

```python
with open("the-verdict.txt", "r", encoding="utf-8") as f:
  raw_text = f.read()

raw_text
```

```python
import re

text = "Hello, world. This is a test."

re_ptrn_for_separating_punt = '([,.:;?"()\']|--|\s)'

result = re.split(re_ptrn_for_separating_punt, raw_text)

result
```

```python
# Strip whitespace

preprocessed = [item for item in result if item.strip()]
len(preprocessed)

preprocessed
```

## **Vocabulary**

```python
# Sort the and remove the duplicates to create the vocab

preprocessed = sorted(set(preprocessed))

vocab_size = len(preprocessed)

preprocessed
```

```python
# Map tokens to integer

text_to_int = {token: integer for integer, token in enumerate(preprocessed)}

int_to_text = {integer: token for integer, token in enumerate(preprocessed)}
```
